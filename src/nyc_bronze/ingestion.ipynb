```python
(spark.readStream.format("cloudFiles")
  .option("cloudFiles.format", "parquet")
  .option("cloudFiles.schemaEvolutionMode","addNewColumnsWithTypeWidening")
  # The schema location directory keeps track of your data schema over time
  .option("cloudFiles.schemaLocation", "/Volumes/nyc_taxi/bronze/raw_data/_schema")
  .load("/Volumes/nyc_taxi/bronze/raw_data/")
  .writeStream
  .option("mergeSchema","true").option("checkpointLocation", "/Volumes/nyc_taxi/bronze/raw_data/_checkpoint").trigger(availableNow=True)
  .toTable("nyc_taxi.bronze.green_taxi")
)
```
