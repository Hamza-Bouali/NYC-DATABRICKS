```python
from datetime import datetime

import pyspark.sql.functions as F

current_timestamp = datetime.now().strftime("%Y_%m_%d_%H_%M")
quarantine_schema = "nyc_taxi.quarantine"
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {quarantine_schema}")

(
    spark.read.table("nyc_taxi.bronze.green_taxi")
    .filter(F.col("trip_distance") == 0)
    .write.mode("overwrite")
    .saveAsTable(f"{quarantine_schema}.taxi_zero_miles_{current_timestamp}")
)
(
    spark.read.table("nyc_taxi.bronze.green_taxi")
    .filter(F.col("trip_distance") == 0)
    .write.mode("overwrite")
    .saveAsTable(f"{quarantine_schema}.taxi_zero_miles_latest")
)
```
