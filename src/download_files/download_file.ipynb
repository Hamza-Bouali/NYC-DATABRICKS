```python
import os
import requests

# 1. Fetch the job parameters safely from widgets
raw_path = dbutils.widgets.get("raw_path")
year = dbutils.widgets.get("year")
month = dbutils.widgets.get("month")

base_url = dbutils.widgets.get("base_url")

# 2. Construct the exact file path natively 
full_path = f"{raw_path}/green_{month}_{year}.parquet"


url = base_url + f"green_tripdata_{year}-{month}.parquet"
print(f"Downloading file from {url} to {full_path}")
r = requests.get(url)

if r.status_code != 200:
    raise Exception(f"Failed to download file from {url}. Status code: {r.status_code}")

with open(full_path, 'wb') as f:
  f.write(r.content)

print("file downloaded succesfuly")
```
