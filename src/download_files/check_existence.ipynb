```python
import os

# 1. Fetch the job parameters safely from widgets
raw_path = dbutils.widgets.get("raw_path")
year = dbutils.widgets.get("year")
month = dbutils.widgets.get("month")
# 2. Construct the exact file path natively 
full_path = f"{raw_path}/green_{month}_{year}.parquet"

# 3. Check for existence (Supports standard cloud paths or DBFS/UC Volumes)

if os.path.isfile(full_path):
    file_exists = "true"
else:
    file_exists = "false"

# 4. Save the boolean string as a Task Value for the downstream If/Else task
dbutils.jobs.taskValues.set(key="does_file_exist", value=file_exists)
```
